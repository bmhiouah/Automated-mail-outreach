````python
import os
# Read environment variables
base_url = os.getenv("OPENAI_BASE_URL")
api_key = os.getenv("OPENAI_API_KEY")
print(base_url)
print(api_key)
````

````python
import subprocess
from openai import OpenAI

client = OpenAI()

# 1. Charger le contenu du template LaTeX
with open("main.tex", "r", encoding="utf-8") as f:
    latex_template = f.read()

# 2. Demander au LLM de modifier/adapter la partie texte
prompt = f"""
Voici mon CV au format LaTeX :
{latex_template}

rends le plus humble, montre que tu n'as pas un gros ego, en choisissant uniquement les skills les plus simples et pertinent
"""

response = client.chat.completions.create(
    model="deepseek.v3.2",
    messages=[{"role": "user", "content": prompt}]
)

new_latex = response.choices[0].message.content.replace("```latex", "").replace("```", "")

# 3. Sauvegarder le nouveau fichier .tex
with open("cv_genere.tex", "w", encoding="utf-8") as f:
    f.write(new_latex)
````

````python
import subprocess
from pathlib import Path
import shutil

tex_file = Path("main.tex")
build_dir = Path("latex_build")
build_dir.mkdir(exist_ok=True)

subprocess.run([
    "pdflatex",
    "-interaction=nonstopmode",
    "-output-directory", str(build_dir),
    str(tex_file)
], check=True)

shutil.move(
    build_dir / "main.pdf",
    Path("main.pdf")
)

shutil.rmtree(build_dir)
````
