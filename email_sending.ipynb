```python
import smtplib
from email.message import EmailMessage

# Informations du compte
GMAIL = "pro.mhiouahb@gmail.com"
MOT_DE_PASSE_APPLICATION = "cmtj kcmx cqhq pomp"

# Destinataire
DESTINATAIRE = "b.mhiouah@gmail.com"

# Création du mail
message = EmailMessage()
message["From"] = GMAIL
message["To"] = DESTINATAIRE
message["Subject"] = "Mon premier mail avec Python"

message.set_content("""
Bonjour,

Ceci est un email envoyé automatiquement avec Python et Gmail.

Bonne journée !
""")

# Connexion à Gmail et envoi
with smtplib.SMTP_SSL("smtp.gmail.com", 465) as serveur:
    serveur.login(GMAIL, MOT_DE_PASSE_APPLICATION)
    serveur.send_message(message)

print("Email envoyé avec succès !")
```
